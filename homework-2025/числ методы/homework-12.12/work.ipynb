Расстояние между двумя точками на Земле. Сложна, очень сложна


```python
from decimal import Decimal
import math

def earth_distance(lat1, lon1, lat2, lon2):
    R = Decimal('6371.01')
    lat1_r = math.radians(lat1)
    lon1_r = math.radians(lon1)
    lat2_r = math.radians(lat2)
    lon2_r = math.radians(lon2)
    angle = math.acos(
        math.sin(lat1_r) * math.sin(lat2_r) + 
        math.cos(lat1_r) * math.cos(lat2_r) * 
        math.cos(lon1_r - lon2_r)
    )
    
    return R * Decimal(angle)

# Тест 1
dist1 = earth_distance(55, 37, 59, 30)
print(f"Расстояние 1: {dist1.quantize(Decimal('0.01'))} км")
print(f"Ожидалось: 613.83 км")
print(f"Совпадает: {dist1.quantize(Decimal('0.01')) == Decimal('613.83')}")

print()

# Тест 2
dist2 = earth_distance(42, 89, 13, 50)
print(f"Расстояние 2: {dist2.quantize(Decimal('0.01'))} км")
print(f"Ожидалось: 4948.17 км")
print(f"Совпадает: {dist2.quantize(Decimal('0.01')) == Decimal('4948.17')}")
```

Погрешность №1

```python
import decimal
x = 1 / 3
x_star = 0.333
abs_error = abs(x - x_star)
rel_error = abs_error / abs(x)
diff = abs(x - x_star)
diff_str = format(diff, '.20f')
decimal.getcontext().prec = 30

print(f"Абсолютная погрешность: {abs_error}")
print(f"Относительная погрешность: {rel_error}")
print(f"Разность (точная): {diff}")
print(f"Разность (форматированная): {diff_str}")
print("\nОпределение верных цифр в широком смысле:")
m = -3
if abs_error <= 0.5 * 10**m:
    print(f"Все цифры до {abs(m)} знака после запятой верны.")
else:
    print("Не все цифры верны.")

print(f"\nИтог:")
print(f"x = {x}")
print(f"x* = {x_star}")
print(f"Абсолютная погрешность Δ = {abs_error:.2e}")
print(f"Относительная погрешность δ = {rel_error:.2e}")
print(f"Верные знаки: все три цифры '333' (т.е. три знака после запятой) верны в широком смысле.")   
```

Погрешность №2


```python
import fractions

x = fractions.Fraction(1, 3)
y = fractions.Fraction(7, 18)

x_star = 0.333
y_star = 0.389

delta_x = abs(float(x) - x_star)
delta_y = abs(float(y) - y_star)

rel_x = delta_x / abs(float(x))
rel_y = delta_y / abs(float(y))

print(f"x = {x} ≈ {float(x):.10f}")
print(f"y = {y} ≈ {float(y):.10f}")
print(f"x* = {x_star}")
print(f"y* = {y_star}")
print(f"Δx = {delta_x:.2e}")
print(f"Δy = {delta_y:.2e}")
print(f"δx = {rel_x:.2e}")
print(f"δy = {rel_y:.2e}\n")

S = x + y
S_star = x_star + y_star
delta_S = delta_x + delta_y
print(f"Точная сумма S = {S} ≈ {float(S):.10f}")
print(f"Приближённая сумма S* = {S_star:.3f}")
print(f"Оценка ΔS ≤ Δx + Δy = {delta_S:.2e}")
print(f"Фактическая ΔS_факт = {abs(float(S) - S_star):.2e}\n")

D = x - y
D_star = x_star - y_star
delta_D = delta_x + delta_y
print(f"Точная разность D = {D} ≈ {float(D):.10f}")
print(f"Приближённая разность D* = {D_star:.3f}")
print(f"Оценка ΔD ≤ Δx + Δy = {delta_D:.2e}")
print(f"Фактическая ΔD_факт = {abs(float(D) - D_star):.2e}\n")

P = x * y
P_star = x_star * y_star
rel_P = rel_x + rel_y
delta_P = float(abs(P)) * rel_P
print(f"Точное произведение P = {P} ≈ {float(P):.10f}")
print(f"Приближённое произведение P* = {P_star:.5f}")
print(f"Оценка δP ≈ δx + δy = {rel_P:.2e}")
print(f"Оценка ΔP ≈ |P| * δP = {delta_P:.2e}")
print(f"Фактическая δP_факт = {abs(float(P) - P_star)/abs(float(P)):.2e}")
print(f"Фактическая ΔP_факт = {abs(float(P) - P_star):.2e}\n")

Q = x / y
Q_star = x_star / y_star
rel_Q = rel_x + rel_y
delta_Q = float(abs(Q)) * rel_Q
print(f"Точное частное Q = {Q} ≈ {float(Q):.10f}")
print(f"Приближённое частное Q* = {Q_star:.5f}")
print(f"Оценка δQ ≈ δx + δy = {rel_Q:.2e}")
print(f"Оценка ΔQ ≈ |Q| * δQ = {delta_Q:.2e}")
print(f"Фактическая δQ_факт = {abs(float(Q) - Q_star)/abs(float(Q)):.2e}")
print(f"Фактическая ΔQ_факт = {abs(float(Q) - Q_star):.2e}")
```
