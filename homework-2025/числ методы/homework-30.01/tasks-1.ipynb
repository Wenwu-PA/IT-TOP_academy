```python
import numpy as np

def monte_carlo_circle(r, n=1000000):
    points = np.random.uniform(-r, r, (n, 2))
    distances = np.sqrt(points[:, 0]**2 + points[:, 1]**2)
    inside = distances <= r
    return (2*r)**2 * np.sum(inside) / n

radii = [1, 2, 3, 4, 5]
print("Площадь круга:")
print("R   Точная   Монте-Карло  Коэффициент")

for r in radii:
    exact = np.pi * r**2
    mc = monte_carlo_circle(r)
    coeff = mc / r**2
    print(f"{r}   {exact:.4f}   {mc:.4f}       {coeff:.4f}")

avg_coeff = np.mean([monte_carlo_circle(r)/r**2 for r in radii])
print(f"\nПриближение: S ≈ {avg_coeff:.4f} × R²")
```

```python
import numpy as np

def calc_volume(r, n=1000000):
    points = np.random.uniform(-r, r, (n, 3))
    distances = np.sqrt(points[:, 0]**2 + points[:, 1]**2 + points[:, 2]**2)
    inside = distances <= r
    return (2*r)**3 * np.sum(inside) / n

print("Объем шара:")
print("R   Точный   Монте-Карло  Коэффициент")

for r in radii:
    exact = (4/3) * np.pi * r**3
    mc = calc_volume(r)
    coeff = mc / r**3
    print(f"{r}   {exact:.4f}   {mc:.4f}       {coeff:.4f}")

avg_coeff = np.mean([calc_volume(r)/r**3 for r in radii])
print(f"\nПриближение: V ≈ {avg_coeff:.4f} × R³")
```

```python
import numpy as np
import matplotlib.pyplot as plt

sigmoid = lambda x: 1 / (1 + np.exp(-x))
x = np.linspace(-10, 10, 1000)

plt.figure(figsize=(8, 6))
plt.plot(x, sigmoid(x), 'b-', linewidth=3)
plt.title('Сигмоидная функция')
plt.xlabel('x')
plt.ylabel('σ(x)')
plt.grid(True)
plt.show()
```
