```python
import tkinter as tk
from tkinter import messagebox

class HouseDrawer:
    def __init__(self, root):
        self.root = root
        self.root.title("Домик - изменение координат")
        self.root.geometry("800x700")
        
        # Опорная точка
        self.reference_point = (300, 300)
        self.house_width = 120
        self.house_height = 100
        
        # Создание интерфейса
        self.create_widgets()
        
        # Холст для рисования
        self.canvas = tk.Canvas(root, bg="white", width=700, height=500)
        self.canvas.pack(pady=10)
        
        # Начальная отрисовка
        self.draw_house()
    
    def create_widgets(self):
        """Создание элементов интерфейса"""
        # Заголовок
        title_label = tk.Label(self.root, text="Рисование домика с изменяемыми координатами", 
                              font=("Arial", 14, "bold"))
        title_label.pack(pady=10)
        
        # Фрейм для элементов управления
        control_frame = tk.Frame(self.root)
        control_frame.pack(pady=10)
        
        # Инструкция
        instruction = tk.Label(control_frame, text="Введите координаты опорной точки (центра основания домика):")
        instruction.grid(row=0, column=0, columnspan=4, pady=5)
        
        # Ввод координат
        tk.Label(control_frame, text="Координата X:").grid(row=1, column=0, padx=5, pady=10)
        self.x_entry = tk.Entry(control_frame, width=10)
        self.x_entry.grid(row=1, column=1, padx=5, pady=10)
        self.x_entry.insert(0, str(self.reference_point[0]))
        
        tk.Label(control_frame, text="Координата Y:").grid(row=1, column=2, padx=5, pady=10)
        self.y_entry = tk.Entry(control_frame, width=10)
        self.y_entry.grid(row=1, column=3, padx=5, pady=10)
        self.y_entry.insert(0, str(self.reference_point[1]))
        
        # Кнопки
        button_frame = tk.Frame(self.root)
        button_frame.pack(pady=10)
        
        update_btn = tk.Button(button_frame, text="Нарисовать домик", command=self.update_coordinates,
                              bg="lightblue", font=("Arial", 10))
        update_btn.pack(side=tk.LEFT, padx=5)
        
        clear_btn = tk.Button(button_frame, text="Очистить холст", command=self.clear_canvas)
        clear_btn.pack(side=tk.LEFT, padx=5)
        
        # Подсказка
        hint = tk.Label(self.root, text="Подсказка: X: 100-600, Y: 100-400", font=("Arial", 9), fg="gray")
        hint.pack(pady=5)
    
    def update_coordinates(self):
        """Обновление координат опорной точки"""
        try:
            x = int(self.x_entry.get())
            y = int(self.y_entry.get())
            
            # Проверка на границы холста
            if x < 100 or x > 600:
                messagebox.showwarning("Внимание", 
                                      "Координата X должна быть в пределах 100-600")
                return
            if y < 100 or y > 400:
                messagebox.showwarning("Внимание", 
                                      "Координата Y должна быть в пределах 100-400")
                return
                
            self.reference_point = (x, y)
            self.draw_house()
        except ValueError:
            messagebox.showerror("Ошибка", "Введите целые числа для координат!")
    
    def clear_canvas(self):
        """Очистка холста"""
        self.canvas.delete("all")
    
    def draw_house(self):
        """Рисование домика"""
        self.clear_canvas()
        x, y = self.reference_point
        
        # Основание дома (квадрат)
        base_points = [
            x - self.house_width//2, y,  # левый нижний
            x - self.house_width//2, y - self.house_height,  # левый верхний
            x + self.house_width//2, y - self.house_height,  # правый верхний
            x + self.house_width//2, y,  # правый нижний
        ]
        
        # Рисование основания
        self.canvas.create_polygon(base_points, fill="#87CEEB", outline="black", width=2)
        
        # Крыша (треугольник)
        roof_points = [
            x - self.house_width//2 - 10, y - self.house_height,  # левый конец
            x, y - self.house_height - 70,  # верх крыши
            x + self.house_width//2 + 10, y - self.house_height,  # правый конец
        ]
        
        self.canvas.create_polygon(roof_points, fill="#8B4513", outline="black", width=2)
        
        # Дверь
        door_width = 35
        door_height = 60
        self.canvas.create_rectangle(
            x - door_width//2, y - door_height, 
            x + door_width//2, y, 
            fill="#8B4513", outline="black", width=2
        )
        
        # Ручка двери
        self.canvas.create_oval(
            x + door_width//2 - 8, y - door_height//2,
            x + door_width//2 - 4, y - door_height//2 + 4,
            fill="gold", outline="black", width=1
        )
        
        # Окно слева
        self.canvas.create_rectangle(
            x - self.house_width//2 + 15, y - self.house_height//2 - 15,
            x - self.house_width//2 + 45, y - self.house_height//2 + 15,
            fill="#FFFACD", outline="black", width=2
        )
        
        # Окно справа
        self.canvas.create_rectangle(
            x + self.house_width//2 - 45, y - self.house_height//2 - 15,
            x + self.house_width//2 - 15, y - self.house_height//2 + 15,
            fill="#FFFACD", outline="black", width=2
        )
        
        # Крестики на окнах
        self.canvas.create_line(
            x - self.house_width//2 + 30, y - self.house_height//2 - 15,
            x - self.house_width//2 + 30, y - self.house_height//2 + 15,
            fill="black", width=1
        )
        self.canvas.create_line(
            x - self.house_width//2 + 15, y - self.house_height//2,
            x - self.house_width//2 + 45, y - self.house_height//2,
            fill="black", width=1
        )
        
        self.canvas.create_line(
            x + self.house_width//2 - 30, y - self.house_height//2 - 15,
            x + self.house_width//2 - 30, y - self.house_height//2 + 15,
            fill="black", width=1
        )
        self.canvas.create_line(
            x + self.house_width//2 - 45, y - self.house_height//2,
            x + self.house_width//2 - 15, y - self.house_height//2,
            fill="black", width=1
        )
        
        # Труба
        self.canvas.create_rectangle(
            x + self.house_width//4, y - self.house_height - 70,
            x + self.house_width//4 + 15, y - self.house_height - 40,
            fill="gray", outline="black", width=1
        )
        
        # Дым из трубы
        self.canvas.create_oval(
            x + self.house_width//4 + 5, y - self.house_height - 75,
            x + self.house_width//4 + 20, y - self.house_height - 65,
            fill="lightgray", outline="lightgray"
        )
        
        # Трава перед домом
        self.canvas.create_rectangle(
            0, y, 700, 500,
            fill="#90EE90", outline=""
        )
        
        # Текст с координатами
        coord_text = f"Координаты центра: ({x}, {y})"
        self.canvas.create_text(x, y + 40, text=coord_text, font=("Arial", 10), fill="darkblue")

def main():
    root = tk.Tk()
    app = HouseDrawer(root)
    root.mainloop()

if __name__ == "__main__":
    main()
```
