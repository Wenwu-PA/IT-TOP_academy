домик

```python
import tkinter as tk
from tkinter import simpledialog

def draw_house(x, y, size=50):
    canvas.delete("all")
    canvas.create_rectangle(x, y, x+size, y+size, fill='green', outline='black')
    canvas.create_polygon(x-size//4, y, 
                         x+size//2, y-size//2, 
                         x+size+size//4, y, 
                         fill='blue', outline='black')
    window_width = size // 2.5
    window_height = size // 1.8 
    window_x = x + (size - window_width) // 2
    window_y = y + (size - window_height) // 3
    canvas.create_rectangle(window_x, window_y, 
                           window_x + window_width, window_y + window_height, 
                           fill='black', outline='white', width=3)

def update_coordinates():
    x = simpledialog.askinteger("Координата X", "Введите координату X:", parent=root)
    y = simpledialog.askinteger("Координата Y", "Введите координату Y:", parent=root)
    
    if x is not None and y is not None:
        draw_house(x, y)
root = tk.Tk()
root.title("Домик")
canvas = tk.Canvas(root, width=800, height=600, bg='white')
canvas.pack()
btn = tk.Button(root, text="Изменить координаты", command=update_coordinates)
btn.pack()

draw_house(100, 200)
root.mainloop()
```

лягушка

```python
import tkinter as tk
from tkinter import simpledialog

def draw_frog(x, y):
    canvas.delete("all")
    mouth_width = 100
    mouth_height = 60
    eye_radius = 20

    upper_points = [
        x - mouth_width, y,
        x + mouth_width, y,
        x, y - mouth_height
    ]
    lower_points = [
        x - mouth_width, y,
        x + mouth_width, y,
        x, y + mouth_height
    ]
    canvas.create_polygon(upper_points, fill='blue', outline='black', width=2)
    canvas.create_polygon(lower_points, fill='green', outline='black', width=2)
    canvas.create_oval(
        x - mouth_width - eye_radius, 
        y - eye_radius,
        x - mouth_width + eye_radius, 
        y + eye_radius,
        fill='red', outline='black', width=2
    )
    canvas.create_oval(
        x + mouth_width - eye_radius, 
        y - eye_radius,
        x + mouth_width + eye_radius, 
        y + eye_radius,
        fill='yellow', outline='black', width=2
    )

def update_coordinates():
    x = simpledialog.askinteger("Координата X", "Введите координату X:", parent=root)
    y = simpledialog.askinteger("Координата Y", "Введите координату Y:", parent=root)
    if x is not None and y is not None:
        draw_frog(x, y)

root = tk.Tk()
root.title("Простая лягушка")
canvas = tk.Canvas(root, width=800, height=600)
canvas.pack()
btn = tk.Button(root, text="Изменить координаты", command=update_coordinates)
btn.pack(pady=10)
draw_frog(400, 300)

root.mainloop()
```

```python
import tkinter as tk
from tkinter import simpledialog

def draw_crown(x, y):
    canvas.delete("all")
    crown_width = 150
    crown_height = 80
    crown_points = [
        x - crown_width, y,
        x - crown_width, y - crown_height * 1.3,
        x - crown_width * 0.8, y - crown_height,
        x - crown_width * 0.4, y - crown_height * 0.5,
        x, y - crown_height * 1.3,
        x + crown_width * 0.4, y - crown_height * 0.5,
        x + crown_width * 0.8, y - crown_height,
        x + crown_width, y - crown_height * 1.3,
        x + crown_width, y,
        x - crown_width, y,
    ]
    canvas.create_polygon(crown_points, fill='gold', outline='black', width=2)
    circle_radius = 12
    canvas.create_oval(
        x - crown_width - circle_radius,
        y - crown_height * 1.3 - circle_radius,
        x - crown_width + circle_radius,
        y - crown_height * 1.3 + circle_radius,
        fill='yellow', outline='black', width=2
    )
    canvas.create_oval(
        x - circle_radius,
        y - crown_height * 1.3 - circle_radius,
        x + circle_radius,
        y - crown_height * 1.3 + circle_radius,
        fill='yellow', outline='black', width=2
    )
    canvas.create_oval(
        x + crown_width - circle_radius,
        y - crown_height * 1.3 - circle_radius,
        x + crown_width + circle_radius,
        y - crown_height * 1.3 + circle_radius,
        fill='yellow', outline='black', width=2
    )
def update_coordinates():
    x = simpledialog.askinteger("X", "Введите X:", parent=root)
    y = simpledialog.askinteger("Y", "Введите Y:", parent=root)
    if x is not None and y is not None:
        draw_crown(x, y)

root = tk.Tk()
root.title("Я устал босс, я сделал все что смог")
canvas = tk.Canvas(root, width=800, height=600)
canvas.pack()
tk.Button(root, text="Изменить координаты", command=update_coordinates).pack(pady=10)
draw_crown(400, 300)
root.mainloop()
```

```python
import tkinter as tk

def animate_squares():
    global square1_id, square2_id, x1_pos, x2_pos
    canvas.move(square1_id, 5, 0)
    x1_pos += 5
    canvas.move(square2_id, -5, 0)
    x2_pos -= 5
    if x1_pos > 850:
        canvas.coords(square1_id, -50, 200, 0, 250)
        x1_pos = -50
    if x2_pos < -50:
        canvas.coords(square2_id, 800, 350, 850, 400)
        x2_pos = 800
    
    root.after(30, animate_squares)
root = tk.Tk()
canvas = tk.Canvas(root, width=800, height=600)
canvas.pack()
square_size = 50
x1_pos = -50
square1_id = canvas.create_rectangle(
    x1_pos, 200,
    x1_pos + square_size, 250,
    fill='blue',
    outline='black',
    width=2,
    tags='square1'
)

x2_pos = 800
square2_id = canvas.create_rectangle(
    x2_pos, 350,
    x2_pos + square_size, 400,
    fill='red',
    outline='black',
    width=2,
    tags='square2'
)
animate_squares()
root.mainloop()
```

```python
import tkinter as tk

def animate():
    global speed1, speed2
    x1, y1, x2, y2 = canvas.coords(square1)
    x3, y3, x4, y4 = canvas.coords(square2)
    ox1, oy1, ox2, oy2 = canvas.coords(obstacle)
    if speed1 > 0 and x2 >= ox1 and x1 < ox2:
        speed1 = -speed1
        canvas.coords(square1, ox1 - 50, y1, ox1, y2)
    elif speed1 < 0 and x1 <= ox2 and x2 > ox1:
        speed1 = -speed1
        canvas.coords(square1, ox2, y1, ox2 + 50, y2)
    elif speed2 > 0 and x4 >= ox1 and x3 < ox2:
        speed2 = -speed2
        canvas.coords(square2, ox1 - 50, y3, ox1, y4)
    elif speed2 < 0 and x3 <= ox2 and x4 > ox1:
        speed2 = -speed2
        canvas.coords(square2, ox2, y3, ox2 + 50, y4)
    elif x2 >= 800:
        speed1 = -abs(speed1)
        canvas.coords(square1, 750, y1, 800, y2)
    elif x1 <= 0:
        speed1 = abs(speed1)
        canvas.coords(square1, 0, y1, 50, y2)  
    elif x4 >= 800:
        speed2 = -abs(speed2)
        canvas.coords(square2, 750, y3, 800, y4)
    elif x3 <= 0:
        speed2 = abs(speed2)
        canvas.coords(square2, 0, y3, 50, y4)
    canvas.move(square1, speed1, 0)
    canvas.move(square2, speed2, 0)
    root.after(20, animate)
root = tk.Tk()
root.title("Квадраты отталкиваются от стенок")
root.geometry("800x400")
canvas = tk.Canvas(root, width=800, height=400, bg='white')
canvas.pack()
obstacle = canvas.create_rectangle(300, 100, 500, 300, 
                                  fill='lightblue', outline='blue', width=2)
square1 = canvas.create_rectangle(50, 150, 100, 200, fill='red')
square2 = canvas.create_rectangle(700, 150, 750, 200, fill='green')
speed1 = 3
speed2 = -3
animate()

root.mainloop()
```
