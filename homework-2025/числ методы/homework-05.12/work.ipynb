Упражнения: понимание синтаксиса 𝒍𝒂𝒎𝒃𝒅𝒂


```python
from functools import reduce

def char_count_with_reduce(s):
    chars = list(s)
    result = reduce(
        lambda acc, char: {**acc, char: acc.get(char, 0) + 1},chars,{}
    )
    
    return result

text = input("Введите строку: ") #hello world
result = char_count_with_reduce(text)
print(result)  
# {'h': 1, 'e': 1, 'l': 3, 'o': 2, ' ': 1, 'w': 1, 'r': 1, 'd': 1}. 
# Что интересно, за строку он и пробел считает...
```

Сложная сортировка


```python
people = [
    {'name': 'Gosha', 'age': 17},
    {'name': 'Sevo', 'age': 17},
    {'name': 'Leo', 'age': 18},
    {'name': 'Nikita', 'age': 18},
    {'name': 'Arkadiy', 'age': 37}
]

sorted_people = sorted(people, key=lambda x: (-x['age'], x['name']))

for person in sorted_people:
    print(f"{person['name']}: {person['age']}")
```
