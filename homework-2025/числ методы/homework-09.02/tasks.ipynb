Основы символьных выражений


```python
import sympy as sp

x, y, z = sp.symbols('x y z')

expr = x**2 + y**2 + z**2

print("Тип выражения:", type(expr))
print("Выражение expr:", expr)
```

Раскрытие скобок


```python
import sympy as sp

x, y = sp.symbols('x y')
expr = (x + y)**3 * (x - y)**2
expanded_expr = sp.expand(expr)

print("Исходное выражение:", expr)
print("После раскрытия:", expanded_expr)
```

Разложение на множители


```python
import sympy as sp

x = sp.symbols('x')
expr = x**4 - 5*x**3 + 5*x**2 - 5*x + 4
factored_expr = sp.factor(expr)

print("Исходное выражение:", expr)
print("После факторизации:", factored_expr)
```

Упрощение и подстановка


```python
import sympy as sp
#1
x = sp.symbols('x')
expr = (x**2 - 4) / (x - 2)

print("1) Исходное выражение:", expr)

#2
simplified_expr = sp.simplify(expr)
print("2) После упрощения:", simplified_expr)

#3
value_at_5 = simplified_expr.subs(x, 5)
print("3) Значение при x = 5:", value_at_5)

print(f"Оригинальное выражение при x=5: {expr.subs(x, 5)}")
print(f"Упрощенное выражение при x=5: {value_at_5}")
```

Комбинированное использование


```python
import sympy as sp
A, B = sp.symbols('A B')
S = A**2*B + 2*A*B**2 + B**3
H = A - B
V = S * H

print("1) Символы объявлены: A, B")
print("2) Площадь основания S =", S)
print("   Высота H =", H)
print("   Объем V = S * H =", V)
expanded_V = sp.expand(V)
print("\n3) Раскрытый объем (полный многочлен):")
print("   V =", expanded_V)
factored_V = sp.factor(expanded_V)
print("\n4) После факторизации:")
print("   V =", factored_V)

print("\nПроверка:")
print("   Исходное V (S * H):", V)
print("   После раскрытия и факторизации:", factored_V)
print("   Совпадают ли выражения?", V.equals(factored_V))
```
