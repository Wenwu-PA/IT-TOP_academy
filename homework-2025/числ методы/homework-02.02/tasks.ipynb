1. plot() - линейный график

Параметры персонализации:
- color: цвет линии/маркера
- linestyle: стиль линии ('-', '--', ':', '-.')

- marker: тип маркера ('o', 's', '^', 'D', '*')
- markersize: размер маркера
- label: метка для легенды
- alpha: прозрачность

````python
import numpy as np
import matplotlib.pyplot as plt

VARIANT = int(input("введите вариант задания: "))
functions = {
    1: (lambda x: 5*np.sin(5*x)*x**2 + x**3, [0, 10*np.pi], "5·sin(5x)·x² + x³"),
    2: (lambda x: 5*np.cos(3*x)*x**2 + x**2, [0, 15*np.pi], "5·cos(3x)·x² + x²"),
    3: (lambda x: 5*np.sin(5*x)*x + x**3*np.cos(x), [0, 20*np.pi], "5·sin(5x)·x + x³·cos(x)"),
    4: (lambda x: 2*np.sin(10*x) + x*np.cos(x), [0, 10*np.pi], "2·sin(10x) + x·cos(x)"),
    5: (lambda x: 5*x*np.sin(5*x) + 10*x*np.cos(x), [0, 15*np.pi], "5x·sin(5x) + 10x·cos(x)"),
    6: (lambda x: x**2*np.sin(2*x) + 5*x*np.cos(x**2), [0, 6*np.pi], "x²·sin(2x) + 5x·cos(x²)"),
    7: (lambda x: 5*x**2*np.sin(7*x) - 5*x*np.cos(x), [0, 10*np.pi], "5x²·sin(7x) - 5x·cos(x)"),
    8: (lambda x: 5*x**2*np.cos(5*x) - 5*x**3*np.cos(2*x), [0, 8*np.pi], "5x²·cos(5x) - 5x³·cos(2x)"),
    9: (lambda x: 5*x**2*np.cos(5*x) - 5*x**3*np.sin(3*x), [0, 5*np.pi], "5x²·cos(5x) - 5x³·sin(3x)"),
    10: (lambda x: np.cos(5*x**2) - 5*x**2*np.cos(x), [0, 5*np.pi], "cos(5x²) - 5x²·cos(x)"),
    11: (lambda x: 5*x*np.sin(5*x**2) + x**2*np.sin(3*x), [0, 5*np.pi], "5x·sin(5x²) + x²·sin(3x)"),
    12: (lambda x: 2*x**2*np.exp(5*x) - 3*x**3*np.cos(2*x), [0, 2*np.pi], "2x²·exp(5x) - 3x³·cos(2x)"),
    13: (lambda x: 5*x**2*np.cos(5*x) - 5*np.exp(2*x), [0, 2*np.pi], "5x²·cos(5x) - 5·exp(2x)"),
    14: (lambda x: 5*x**2*np.cos(x)*np.cos(2*x), [0, 3*np.pi], "5x²·cos(x)·cos(2x)"),
    15: (lambda x: x**2 + np.sin(x)*np.sin(2*x), [0, 3*np.pi], "x² + sin(x)·sin(2x)")
}

func, interval, formula = functions.get(VARIANT, functions[1])
a, b = interval
x = np.linspace(a, b, 500)

def func_def(x):
    return func(x)
y1 = func_def(x)
fig1, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
fig1.patch.set_facecolor('#F0F0F0')
ax1.plot(x, y1, 'b-', linewidth=2, marker='o', markersize=3, label='Через def')
ax1.set_title(f'Вариант {VARIANT}: Через функцию')
ax1.set_xlabel('x')
ax1.set_ylabel('f(x)')
ax1.grid(True)
ax1.legend()
ax2.plot(x, y2, 'r--', linewidth=2, marker='s', markersize=3, label='Прямое выражение')
ax2.set_title(f'Вариант {VARIANT}: Прямое выражение')
ax2.set_xlabel('x')
ax2.set_ylabel('f(x)')
ax2.grid(True)
ax2.legend()
plt.suptitle(f'f(x) = {formula}', fontsize=14)
plt.tight_layout()
plt.show()
fig2, (ax3, ax4) = plt.subplots(2, 1, figsize=(10, 8))
fig2.patch.set_facecolor('#F5F5F5')
ax3.plot(x, y1, 'g-', linewidth=2, marker='^', markersize=4, label='Через def')
ax3.set_title('Верхний: функция через def')
ax3.set_ylabel('f(x)')
ax3.grid(True)
ax3.legend()
ax4.plot(x, y2, 'm-.', linewidth=2, marker='D', markersize=4, label='Прямое выражение')
ax4.set_title('Нижний: прямое выражение')
ax4.set_xlabel('x')
ax4.set_ylabel('f(x)')
ax4.grid(True)
ax4.legend()
plt.suptitle(f'Вариант {VARIANT}: f(x) = {formula}', fontsize=14)
plt.tight_layout()
plt.show()
print(f"Вариант: {VARIANT}")
print(f"Функция: f(x) = {formula}")
print(f"Интервал: [{a:.2f}, {b:.2f}]")
print(f"Максимальная разность: {np.max(np.abs(y1 - y2)):.2e}")
print("Результаты одинаковы (разница в пределах погрешности)")
````

Seaborn 2: Связь между общим счётом и чаевыми

````python
import seaborn as sns
import matplotlib.pyplot as plt
tips = sns.load_dataset('tips')
plt.figure(figsize=(10, 6))
sns.scatterplot(data=tips, x='total_bill', y='tip', 
                hue='smoker',
                style='smoker',
                s=100,
                alpha=0.7,
                palette=['red', 'green'])
plt.xlabel('Общий счёт ($)', fontsize=12)
plt.ylabel('Чаевые ($)', fontsize=12)
plt.grid(True, alpha=0.3, linestyle='--')
sns.regplot(data=tips[tips['smoker'] == 'Yes'], x='total_bill', y='tip', 
            scatter=False, color='red', label='Курящие (тренд)', ci=None)
sns.regplot(data=tips[tips['smoker'] == 'No'], x='total_bill', y='tip', 
            scatter=False, color='green', label='Некурящие (тренд)', ci=None)
plt.legend(title='Курящий посетитель', fontsize=11, title_fontsize=12)

plt.tight_layout()
plt.show()
print("=" * 60)
print("АНАЛИЗ СВЯЗИ МЕЖДУ СЧЁТОМ И ЧАЕВЫМИ")
print("=" * 60)
print("\nСтатистика по курящим посетителям:")
print("-" * 40)
for smoker_status in ['Yes', 'No']:
    group_data = tips[tips['smoker'] == smoker_status]
    print(f"\n{smoker_status} (курящий):")
    print(f"  Количество наблюдений: {len(group_data)}")
    print(f"  Средний счёт: ${group_data['total_bill'].mean():.2f}")
    print(f"  Средние чаевые: ${group_data['tip'].mean():.2f}")
    print(f"  % чаевых от счёта: {(group_data['tip'].mean() / group_data['total_bill'].mean() * 100):.1f}%")
    correlation = group_data['total_bill'].corr(group_data['tip'])
    print(f"  Корреляция счёт-чаевые: {correlation:.3f}")

print("\n" + "=" * 60)
print("ВЫВОДЫ:")
print("=" * 60)
print("1. Существует положительная связь между размером счёта и чаевыми")
print("2. Курящие и некурящие посетители показывают схожую зависимость")
print("3. Диаграмма позволяет визуально оценить распределение данных")
print("4. Линии тренда показывают общую тенденцию для каждой группы")
print("=" * 60)
````

Seaborn 3: Средний счёт по дням недели

````python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

plt.figure(figsize=(10, 6))
sns.barplot(data=tips, x='day', y='total_bill')

plt.title('Средний общий счёт по дням недели')
plt.xlabel('День недели')
plt.ylabel('Средний счёт ($)')

ax = plt.gca()
for bar in ax.patches:
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2., height + 0.5,
            f'${height:.2f}', ha='center', va='bottom', fontsize=10)

plt.tight_layout()
plt.show()

print("=" * 50)
print("СТАТИСТИКА:")
print("=" * 50)
for day in tips['day'].unique():
    day_data = tips[tips['day'] == day]['total_bill']
    mean_val = day_data.mean()
    count = len(day_data)
    print(f"{day}: Средний счёт = ${mean_val:.2f}, Наблюдений = {count}")
````
