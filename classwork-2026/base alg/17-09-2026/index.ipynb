```python
n, k = map(int, input().split())

len = [int(input()) for _ in range(n)]

l = 1
r = max(len)
answer = 0

while l <= r:
    mid = (l + r) // 2

    count = 0

    for length in len:
        count += length // mid

    if count >= k:
        answer = mid
        l = mid + 1
    else:
        r = mid - 1

print(answer)
```
