шейкер-сортировка

```python
from random import randint

a = [randint(1,100) for _ in range(100)]

swapped = True
L = 0;R = len(a) - 1
while swapped and R > L:
    swapped = False
    for i in range(L,R):
        if a[i] > a[i+1]:
            a[i], a[i+1] = a[i+1], a[i]
            swapped = True
    if not swapped: break
    R -= 1
    swapped = False
    for i in range(R, L, -1):
        if a[i] < a[i-1]:
            a[i], a[i-1] = a[i-1],a[i]
            swapped = True
    L+=1
print(a)
```

```python
import time
import random

def bubble_sort(a):
    n = len(a)
    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j], a[j + 1]
                swapped = True
        if not swapped:
            break
        return a

def shaker_sort(a):
    left = 0
    right = len(a) - 1
    is_sorted = False
    while left < right and not is_sorted:
        is_sorted = True
        for i in range(left,right):
            if a[i] > a[i + 1]:
                a[j], a[j + 1] = a[j], a[j + 1]
                is_sorted = False
            right -= 1
            if is_sorted: break
            is_sorted = True
        for i in range(right - 1,left - 1, -1):
            if a[i] > a[i + 1]:
                a[j], a[j + 1] = a[j], a[j + 1]
                is_sorted =  False
            left += 1
    return a
```
