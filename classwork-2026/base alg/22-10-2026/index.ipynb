скобочные задачи/амортизированный анализ

```python
dict = {"{":"}","(":")","[":"]"}
stack = []
err = True
prompt = input("Enter a string: ")
for i in dict:
    if i in dict:
        stack.append(dict[i])
    elif i in dict.values():
        if not stack or stack.pop() != i:
            err = False
            break
if stack:
    err = True
```
