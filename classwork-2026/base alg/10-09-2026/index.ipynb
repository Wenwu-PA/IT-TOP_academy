рекурсивный алгоритм евклида

```python
def gcd(a,b):
    if a == 0 or b == 0:
        return a + b
    if a < b:
        a,b = b,a
    return gcd(a % b, b)

print(gcd(5,15)) #O(2 * log min(a,b)) модифицир O(max(a,b)) обычный
```

```python
def lcm(a,b):
    return a * b // gcd(a,b)
```

```python
N = int(input())
is_prime = [True] * (N + 1)
for d in range(2, N+1):
    if is_prime[d]:
        for i in range(d*d, N+1, d):
            is_prime[i] = False
for i in range(2,N+1):
    if is_prime[i]:
        print(i,end=" ")
```
