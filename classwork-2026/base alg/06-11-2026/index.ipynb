```python

from collections import Counter

with open("the-zen-of-python.txt", "r") as file:
    text = file.read().lower()

letters = [char for char in text if char.isalpha()]

counter = Counter(letters)

print(counter)
```

нахождение A число, наибольшее число раз

```python

from collections import Counter

A = list(map(int, input().split()))

counter = Counter(A)
max_count = max(counter.values())

result = max(num for num in counter if counter[num] == max_count)

print(result)
```

```python

n, k = map(int, input().split())
a = list(map(int, input().split()))

for i in range(1, k + 1):
    max_index = 0

    for j in range(1, len(a)):
        if a[j] > a[max_index]:
            max_index = j

    result = a.pop(max_index)

print(result)
```

```python

with open('26-k1.txt') as f:
    n, k = map(int, f.readline().split())
    a = [int(f.readline()) for i in range(n)]

a.sort(reverse=True)

discount = sum(a[:k]) * 0.2
max_price = a[k]

print(max_price, round(discount))
```

```python

with open('26-k3.txt') as f:
    n, k, m = map(int, f.readline().split())
    a = [int(f.readline()) for i in range(n)]

a.sort(reverse=True)

print(a[k + m - 1], a[k - 1])
```

```python

with open('26-k5.txt') as f:
    n, k, m = map(int, f.readline().split())
    a = [int(f.readline()) for i in range(n)]

a.sort()

premium = a[-m]
budget = sum(a[:k]) // k

print(premium, budget)
```

```python

with open('26-k2.txt') as f:
    n, k = map(int, f.readline().split())
    a = [int(f.readline()) for i in range(n)]

a.sort()

a = a[k:n-k]

print(a[-1], sum(a) // len(a))
```
