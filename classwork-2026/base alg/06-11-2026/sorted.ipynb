## 1. Что такое сортировка

**Сортировка** — расстановка элементов массива (или другой коллекции) в заранее заданном порядке.

Порядок может быть разным:

- **По возрастанию:** $1, 3, 4, 5, 21, 45, 179$.
- **По убыванию:** $179, 45, 21, 5, 4, 3, 1$.
- **По свойству:** по последней цифре, сумме цифр, сумме делителей и т. д.
- **Для строк:** в алфавитном / лексикографическом порядке.

Для строк лексикографический порядок зависит от используемого алфавита и правил сравнения. Например, в Python обычное сравнение строк проводится по кодовым точкам Unicode: это не всегда «словарный порядок» конкретного языка.

````python
A = [18, 7, 15, 25, 6]
print("До сортировки:", A)
print("По возрастанию:", sorted(A))
print("По убыванию:", sorted(A, reverse=True))
````

## 2. Что можно сортировать

Чтобы сортировать элементы, нужно определить **отношение порядка** — правило, по которому понятно, какой элемент «меньше», «больше» или равен другому по выбранному ключу.

Для обычного числового порядка выполняется **транзитивность**:

- если $a < b$ и $b < c$, то $a < c$;
- если $a = b$ и $b = c$, то $a = c$.

**Не каждое правило сравнения подходит для сортировки.** Например, правило «камень побеждает ножницы, ножницы — бумагу, бумага — камень» циклично, поэтому общего линейного порядка по этому правилу нет.

В Python для сложных объектов удобно задавать **ключ сортировки** (`key=`): например, возраст пользователя, длину строки или сумму цифр.

````python
words = ["банан", "арбуз", "слива", "киви"]
print("По длине:", sorted(words, key=len))
print("По обычному порядку строк:", sorted(words))
````

## 3. Основная операция — перестановка

В сортировке часто требуется поменять местами два элемента.

С использованием временной переменной:

```python
c = x
x = y
y = c
```

В Python есть более простой способ — **множественное присваивание**:

```python
x, y = y, x
```

Оно работает и с элементами списка: `A[i], A[j] = A[j], A[i]`.

````python
A = [6, 4]
A[0], A[1] = A[1], A[0]
print(A)  # [4, 6]
````

## 4. Характеристики сортировки

| Характеристика | Значение |
|---|---|
| **Устойчивость** | Элементы с равными ключами сохраняют первоначальный взаимный порядок. |
| **Сортировка на месте (in-place)** | Не требует дополнительного массива размером $O(n)$; обычно использует $O(1)$ дополнительной памяти. |
| **Внутренняя** | Все сортируемые данные помещаются в оперативную память. |
| **Внешняя** | Использует внешние носители, когда все данные одновременно не помещаются в ОЗУ. |
| **Адаптивная** | Может работать быстрее, если входные данные уже частично упорядочены. |
| **Неадаптивная** | Обычно выполняет сопоставимый объём работы независимо от первоначального порядка. |

**Важно:** устойчивость имеет значение, когда у разных объектов одинаковый ключ, но сами объекты различаются.

````python
# Два человека имеют одинаковый возраст.
# Устойчивая сортировка по возрасту сохраняет их относительный порядок.
people = [("Аня", 18), ("Борис", 17), ("Вера", 18)]
print(sorted(people, key=lambda person: person[1]))
# Борис, Аня, Вера
````

## 5. Основные алгоритмы сортировки

### Универсальные сортировки

| Алгоритм | Типичное / худшее время | Дополнительная память | Устойчивость |
|---|---|---|---|
| **Пузырьковая** | $O(n^2)$ / $O(n^2)$ | $O(1)$ | Да (при обмене только по строгому `>`) |
| **Выбором** | $O(n^2)$ / $O(n^2)$ | $O(1)$ | Обычно нет |
| **Вставками** | $O(n^2)$ / $O(n^2)$ | $O(1)$ | Да (при стандартной реализации) |
| **Быстрая (quicksort)** | $O(n\log n)$ / $O(n^2)$ | обычно $O(\log n)$, зависит от реализации | Обычно нет |
| **Пирамидальная (heapsort)** | $O(n\log n)$ / $O(n\log n)$ | $O(1)$ | Нет |
| **Слиянием (mergesort)** | $O(n\log n)$ / $O(n\log n)$ | обычно $O(n)$ | Да (при стандартной реализации) |

### Специализированные сортировки

- **Подсчётом:** $O(n+k)$, где $k$ — количество возможных значений ключа или размер их диапазона; требует дополнительной памяти.
- **Поразрядная (radix sort):** порядка $O(d(n+b))$, где $d$ — число разрядов, а $b$ — размер основания / диапазона цифр (в типичной реализации через подсчёт).

В учебной классификации пузырьковую, выбором и вставками объединяют в простые алгоритмы с квадратичным временем. Алгоритмы с $O(n\log n)$ обычно эффективнее на больших массивах.

Обозначение $O(f(n))$ описывает **верхнюю асимптотическую оценку**: как растёт время или расход памяти при увеличении числа элементов.

## 6. Проверка отсортированности массива

Для чисел, отсортированных **по неубыванию**, у каждой соседней пары выполняется:

$$A[i] \le A[i+1]$$

Достаточно пройти по списку один раз: время **$O(n)$**, дополнительная память **$O(1)$**.

````python
def is_sorted(A):
    for i in range(len(A) - 1):
        if A[i] > A[i + 1]:
            return False
    return True

print(is_sorted([1, 2, 3, 4, 6]))  # True
print(is_sorted([1, 3, 2, 4, 6]))  # False
print(is_sorted([]))               # True
````

## 7. «Сортировка бильярдом» / случайная сортировка (Bogosort)

В слайдах приведена идея: **перемешивать массив случайно, пока он не окажется отсортированным**.

Алгоритм наглядный, но непрактичный: число попыток может быть огромным. Для массива из $n$ различных элементов ожидаемое число случайных перестановок порядка $n!$, а вместе с проверкой порядка — порядка $O(n\cdot n!)$ при грубой оценке.

Следующий пример ограничивает число попыток, чтобы ячейка не зависла.

````python
import random

def bogosort_demo(A, max_attempts=10_000):
    A = A.copy()
    for attempt in range(max_attempts + 1):
        if is_sorted(A):
            return A, attempt
        random.shuffle(A)
    return None, max_attempts

random.seed(7)
print(bogosort_demo([3, 1, 2]))
````

## 8. Метод выбора (Selection Sort) — главное

**Идея:** находить минимальный элемент в *ещё не отсортированной* части массива и переставлять его в начало этой части.

Процесс для сортировки по возрастанию:

1. Считаем, что граница отсортированной части находится в позиции `i = 0`.
2. Находим индекс минимального элемента в диапазоне `A[i:]`.
3. Меняем местами `A[i]` и найденный минимум, если это разные позиции.
4. Увеличиваем `i` на 1: слева уже стоит минимальный из оставшихся элементов.
5. Повторяем до предпоследнего элемента. Последний окажется на своём месте автоматически.

**Инвариант алгоритма:** перед каждой итерацией первые `i` элементов стоят в правильном порядке и не больше элементов в оставшемся суффиксе.

В списке из $n$ элементов достаточно **$n-1$ проходов**.

### 8.1. Пример по шагам

Начальный массив со слайда:

`[8, 2, 4, 1, 3, 9, 6, 7]`

| Шаг | Минимум в оставшейся части | Массив после перестановки |
|---:|---:|---|
| 1 | 1 | `[1, 2, 4, 8, 3, 9, 6, 7]` |
| 2 | 2 | `[1, 2, 4, 8, 3, 9, 6, 7]` |
| 3 | 3 | `[1, 2, 3, 8, 4, 9, 6, 7]` |
| 4 | 4 | `[1, 2, 3, 4, 8, 9, 6, 7]` |
| 5 | 6 | `[1, 2, 3, 4, 6, 9, 8, 7]` |
| 6 | 7 | `[1, 2, 3, 4, 6, 7, 8, 9]` |
| 7 | 8 | `[1, 2, 3, 4, 6, 7, 8, 9]` |

### 8.2. Реализация на Python

Внешний цикл определяет позицию, куда нужно поставить минимум. Внутренний ищет минимум **с позиции `i + 1` до конца**, не затрагивая уже отсортированную часть.

````python
def selection_sort(A):
    A = A.copy()                  # не изменяем исходный список
    n = len(A)
    for i in range(n - 1):
        min_index = i            # минимум пока в A[i]
        for j in range(i + 1, n):
            if A[j] < A[min_index]:
                min_index = j
        if min_index != i:
            A[i], A[min_index] = A[min_index], A[i]
    return A

numbers = [8, 2, 4, 1, 3, 9, 6, 7]
print("Исходный:", numbers)
print("Результат:", selection_sort(numbers))
````

> Если требуется **строго на месте**, уберите `A = A.copy()` из функции. Тогда сортировка изменит переданный список непосредственно и дополнительная память станет $O(1)$; в показанном варианте копия намеренно сохраняет оригинал и требует $O(n)$ памяти.

### 8.3. Просмотр промежуточных состояний

````python
def selection_sort_steps(A):
    A = A.copy()
    n = len(A)
    for i in range(n - 1):
        min_index = i
        for j in range(i + 1, n):
            if A[j] < A[min_index]:
                min_index = j
        A[i], A[min_index] = A[min_index], A[i]
        print(f"Шаг {i + 1}: минимум = {A[i]:>2}, массив = {A}")
    return A

selection_sort_steps([8, 2, 4, 1, 3, 9, 6, 7])
````

## 9. Почему пример из презентации не работает

На одном из слайдов приведён код в духе:

```python
A = [1, 21, 3, -46, 53, -6, 117]
N = len(A)
for k in range(N - 1):
    nM = 0
    for i in range(N):
        if A[i] < A[nM]:
            nM = i
    A[k], A[nM] = A[nM], A[k]
```

**Ошибка:** для каждого `k` индекс минимума снова сбрасывается в `0`, а внутренний цикл ищет минимум **во всём массиве**, включая уже обработанную часть. Из-за этого ранее поставленные элементы могут снова менять позиции.

**Исправление:** начинать `min_index = k` и искать только среди индексов от `k + 1` до `N - 1`.

````python
A = [1, 21, 3, -46, 53, -6, 117]
N = len(A)
for k in range(N - 1):
    min_index = k
    for i in range(k + 1, N):
        if A[i] < A[min_index]:
            min_index = i
    if min_index != k:
        A[k], A[min_index] = A[min_index], A[k]

print(A)  # [-46, -6, 1, 3, 21, 53, 117]
````

## 10. Сложность сортировки выбором

**Время:** на первом проходе нужно $n-1$ сравнений, на втором — $n-2$, …, на последнем — 1.

$$T(n) = (n-1) + (n-2) + \ldots + 1 = \frac{n(n-1)}{2}$$

Следовательно, временная сложность — **$\Theta(n^2)$** в лучшем, среднем и худшем случаях.

- Число сравнений: всегда $n(n-1)/2$ в стандартном варианте.
- Число обменов: **не более $n-1$** (если пропускать обмен элемента с самим собой).
- Дополнительная память: **$O(1)$** для алгоритма, работающего на исходном списке.
- **Неадаптивная:** уже отсортированный массив всё равно потребует квадратичного числа сравнений.
- **В общем случае неустойчивая:** перестановка минимума может нарушить порядок элементов с равными ключами.

В демонстрационной функции `selection_sort` сделана копия списка, поэтому *всей функции* требуется $O(n)$ дополнительной памяти, хотя сам алгоритм можно реализовать с $O(1)$.

````python
for n in [5, 10, 100, 1000]:
    comparisons = n * (n - 1) // 2
    print(f"n = {n:4d} → {comparisons:>7d} сравнений")
````

## 11. Метод выбора через последовательные обмены

На последнем слайде показан вариант на C++:

```cpp
void SelectionSort(int A[], int size) {
    for (int i = 0; i < size; i++) {
        for (int j = i + 1; j < size; j++) {
            if (A[i] > A[j]) {
                int tmp = A[i];
                A[i] = A[j];
                A[j] = tmp;
            }
        }
    }
}
```

Он тоже сортирует, но обменивает элементы **каждый раз, когда встречает меньший**, а не один раз после поиска минимума. Такой вариант часто называют сортировкой обменами (*exchange sort*): сравнений $O(n^2)$ и обменов может быть $O(n^2)$, в отличие от классического метода выбора, у которого не более $n-1$ обменов.

````python
def exchange_sort(A):
    A = A.copy()
    n = len(A)
    for i in range(n):
        for j in range(i + 1, n):
            if A[j] < A[i]:
                A[i], A[j] = A[j], A[i]
    return A

print(exchange_sort([4, 3, 1, 2]))  # [1, 2, 3, 4]
````

### Короткая проверка знаний

- **Что такое сортировка?** Расстановка элементов в заданном порядке.
