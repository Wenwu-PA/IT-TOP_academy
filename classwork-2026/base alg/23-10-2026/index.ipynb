Обработка числовых последовательностей

```python
with open("271-A.txt", "r") as f:
    n = int(f.readline())
    a = list(map(int, f.readline().split()))
    r = a.pop()

max_even = -1
max_odd = -1

for x in a:
    if x % 2 == 0:
        max_even = max(max_even, x)
    else:
        max_odd = max(max_odd, x)

if max_even == -1 or max_odd == -1:
    s = -1
else:
    s = max_even + max_odd

print("Контрольная сумма:", s)

if s == r:
    print("Контроль пройден")
else:
    print("Контроль не пройден")
```
