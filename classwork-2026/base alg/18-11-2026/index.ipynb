бин поиск (log(N)) с сортировкой вставок

```python
def binary_insertion_sort(A):
    for i in range(1, len(A)):
        x = A[i]
        left, right = 0, i
        while left < right:
            mid = (left + right) // 2
            if A[mid] < x:
                left = mid + 1
            else:
                right = mid
        A[left + 1 : i + 1] = A[left : i]
        A[left] = x
    return A
```

сортировка вставками с бин поиском

```python
class HeavyObject:
    def __init__(self, value: int, delay_loops: int = 3000):
        self.value = value
        self.delay_loops = delay_loops

    def __lt__(self, other):
        if not isinstance(other, HeavyObject):
            return NotImplemented
        for _ in range(self.delay_loops):
            pass
        return self.value < other.value
```

```python
def insertion_sort(A):
    for i in range(1, len(A)):
        x = A[i]
        j = i - 1
        while j >= 0 and A[j] > x:
            A[j + 1] = A[j]
            j -= 1
        A[j + 1] = x
    return A
```

от вставки к Шеллу

```python
def shell_sort(A):
    n = len(A)
    gap = n // 2
    while gap > 0:
        for i in range(gap, n):
            temp = A[i]
            j = i
            while j >= gap and A[j - gap] > temp:
                A[j] = A[j - gap]
                j -= gap
            A[j] = temp
        gap //= 2
    return A
```

оптимизация Седжвика

```python
def generate_sedgewick_gaps(n):
    gaps = []
    k = 0
    while True:
        gap = 9 * (4 ** k) - 9 * (2 ** k) + 1
        if gap >= n:
            break
        gaps.append(gap)
        k += 1
    k = 1
    while True:
        gap = (4 ** k) - (3 * (2 ** k)) + 1
        if gap >= n:
            break
        gaps.append(gap)
        k += 1
    gaps.sort(reverse=True)
    return gaps
```
