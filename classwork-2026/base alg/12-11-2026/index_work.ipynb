шейкерная реализиция поиска отрезка длины K с максимальной суммой

```python
with open("26-k2.txt") as f:
    N, K = map(int, f.readline().split())
    A = [int(f.readline()) for i in range(N)]

left = 0
right = N - 1

while left < right:
    for i in range(left, right):
        if A[i] > A[i + 1]:
            A[i], A[i + 1] = A[i + 1], A[i]

    right -= 1

    for i in range(right, left, -1):
        if A[i] < A[i - 1]:
            A[i], A[i - 1] = A[i - 1], A[i]

    left += 1

sm = 0
for i in range(K, N - K):
    sm += A[i]

print(max(A[K:N-K]), sm / (N - 2 * K))
```

быстрая сортировка

```python
def qsort(A, l, r, reverse=False):
    if l >= r:
        return

    x = A[(l + r) // 2]
    i = l
    j = r

    while i <= j:
        if not reverse:
            while A[i] < x:
                i += 1
            while A[j] > x:
                j -= 1
        else:
            while A[i] > x:
                i += 1
            while A[j] < x:
                j -= 1

        if i <= j:
            A[i], A[j] = A[j], A[i]
            i += 1
            j -= 1

    if l < j:
        qsort(A, l, j, reverse)
    if i < r:
        qsort(A, i, r, reverse)


A = [5, 3, 4, 2, 1, 6, 3, 2]
n = len(A)
m = n // 2

qsort(A, 0, m - 1)
qsort(A, m, n - 1, True)

print(A)
```

число различных чисел в сортировке

```python
def qsort(A, l, r):
    if l >= r:
        return

    x = A[(l + r) // 2]
    i = l
    j = r

    while i <= j:
        while A[i] < x:
            i += 1
        while A[j] > x:
            j -= 1

        if i <= j:
            A[i], A[j] = A[j], A[i]
            i += 1
            j -= 1

    qsort(A, l, j)
    qsort(A, i, r)


A = [5, 3, 4, 2, 1, 6, 3, 2, 4]

qsort(A, 0, len(A) - 1)

k = 1
for i in range(1, len(A)):
    if A[i] != A[i - 1]:
        k += 1

print("После сортировки:", *A)
print("Различных чисел:", k)
```

бенчмарк (сравнение перестановок)

```python

from random import randint

def bubble(A):
    k = 0
    for i in range(len(A) - 1):
        for j in range(len(A) - i - 1):
            if A[j] > A[j + 1]:
                A[j], A[j + 1] = A[j + 1], A[j]
                k += 1
    return k

def selection(A):
    k = 0
    for i in range(len(A) - 1):
        m = i
        for j in range(i + 1, len(A)):
            if A[j] < A[m]:
                m = j
        if m != i:
            A[i], A[m] = A[m], A[i]
            k += 1
    return k

def qsort(A, l, r):
    k = 0
    i = l
    j = r
    x = A[(l + r) // 2]

    while i <= j:
        while A[i] < x:
            i += 1
        while A[j] > x:
            j -= 1

        if i <= j:
            if i != j:
                A[i], A[j] = A[j], A[i]
                k += 1
            i += 1
            j -= 1

    if l < j:
        k += qsort(A, l, j)
    if i < r:
        k += qsort(A, i, r)

    return k


b = 0
s = 0
q = 0

for t in range(100):
    A = [randint(1, 10000) for i in range(1000)]

    B = A.copy()
    C = A.copy()
    D = A.copy()

    b += bubble(B)
    s += selection(C)
    q += qsort(D, 0, len(D) - 1)

print("Пузырёк:", b / 100)
print("Выбором:", s / 100)
print("Быстрая:", q / 100)
```

слияние массивов

```python

A = list(map(int, input().split()))
B = list(map(int, input().split()))

C = []
i = 0
j = 0

while i < len(A) and j < len(B):
    if A[i] <= B[j]:
        C.append(A[i])
        i += 1
    else:
        C.append(B[j])
        j += 1

while i < len(A):
    C.append(A[i])
    i += 1

while j < len(B):
    C.append(B[j])
    j += 1

print(*C)
```

слияние

```python
def merge(A, B):
    i = 0
    j = 0

    while i < len(A) and j < len(B):
        if A[i]['timestamp'] <= B[j]['timestamp']:
            yield A[i]
            i += 1
        else:
            yield B[j]
            j += 1

    while i < len(A):
        yield A[i]
        i += 1

    while j < len(B):
        yield B[j]
        j += 1


def merge_logs(servers):
    result = []

    for server in servers:
        result = list(merge(result, server))

    return result


server_1 = [{'timestamp': 100, 'tx': 'A'}, {'timestamp': 150, 'tx': 'B'}]
server_2 = [{'timestamp': 110, 'tx': 'C'}, {'timestamp': 160, 'tx': 'D'}]

servers = [server_1, server_2]

for log in merge_logs(servers):
    print(log, end=",\n")
```

очередь заказорв службы доставки

```python

def insert_order(queue, order):
    queue.append(order)
    i = len(queue) - 1

    while i > 0:
        a = queue[i]
        b = queue[i - 1]

        if a[1] < b[1] or (a[1] == b[1] and a[2] < b[2]):
            queue[i], queue[i - 1] = queue[i - 1], queue[i]
            i -= 1
        else:
            break

    return queue


queue = []

insert_order(queue, [101, 2, 120])
insert_order(queue, [102, 1, 150])
insert_order(queue, [103, 2, 100])
insert_order(queue, [104, 1, 90])
insert_order(queue, [105, 3, 80])

for order in queue:
    print(order)
```
