# Сортировка слиянием (Merge Sort)

**Идея:** «разделяй и властвуй». Делим массив пополам, сортируем половины рекурсивно, затем сливаем два отсортированных массива.

## Алгоритм

1. **Divide:** разделить массив на две половины.
2. **Conquer:** рекурсивно отсортировать каждую половину. Массив из одного элемента уже отсортирован.
3. **Combine:** сравнивать первые оставшиеся элементы двух половин и переносить меньший в результат. В конце добавить остаток.

Пример: `[5, 2, 4, 1]` → `[5, 2]` и `[4, 1]` → `[2, 5]` и `[1, 4]` → `[1, 2, 4, 5]`.

## Реализация

```python
def merge(A, B):
    C = []
    i = 0
    j = 0

    while i < len(A) and j < len(B):
        if A[i] <= B[j]:
            C.append(A[i])
            i += 1
        else:
            C.append(B[j])
            j += 1

    C += A[i:] + B[j:]
    return C

def mergeSort(A):
    if len(A) <= 1:
        return A

    m = len(A) // 2
    L = mergeSort(A[:m])
    R = mergeSort(A[m:])
    return merge(L, R)

A = [5, 3, 4, 2, 1, 6, 3, 2, 4]
print(mergeSort(A))
```

## Сложность и свойства

- **Время:** `O(N log N)` в лучшем, среднем и худшем случаях.
- **Дополнительная память:** `O(N)` для слияния (в этой реализации со срезами также создаются временные списки).
- **Устойчивость:** одинаковые элементы сохраняют исходный порядок, если при равенстве выбирать элемент из левой половины (`<=`).
- **Минус:** нужна дополнительная память.

**Отличие от QuickSort:** Merge Sort делит массив пополам и объединяет отсортированные части; QuickSort разделяет элементы относительно опорного значения.

## Задание: слияние логов

Каждый сервер хранит события по возрастанию `timestamp`. Используем два указателя, чтобы объединить логи за линейное время без повторной сортировки.

```python
server_1 = [{'timestamp': 100, 'tx': 'A'}, {'timestamp': 150, 'tx': 'B'}]
server_2 = [{'timestamp': 110, 'tx': 'C'}, {'timestamp': 160, 'tx': 'D'}]

def mergeLogs(A, B):
    C = []
    i = 0
    j = 0
    while i < len(A) and j < len(B):
        if A[i]['timestamp'] <= B[j]['timestamp']:
            C.append(A[i])
            i += 1
        else:
            C.append(B[j])
            j += 1
    C += A[i:] + B[j:]
    return C

for log in mergeLogs(server_1, server_2):
    print(log)
```

## Рекурсивное слияние (из лекции)

Базовый случай — один из массивов пуст. Иначе выбираем меньший первый элемент и рекурсивно сливаем остаток. Такой вариант проще для понимания, но из-за срезов и глубокой рекурсии хуже подходит для больших данных.

```python
def recursivelyMerged(A, B):
    if len(A) == 0:
        return B
    if len(B) == 0:
        return A
    if A[0] <= B[0]:
        return [A[0]] + recursivelyMerged(A[1:], B)
    return [B[0]] + recursivelyMerged(A, B[1:])

print(recursivelyMerged([1, 3, 5], [2, 4, 6]))
```
