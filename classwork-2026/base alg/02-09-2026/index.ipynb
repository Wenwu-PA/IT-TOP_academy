```python
N = int(input())
D = [1,N]
d = 2
while d * d <= N:
    if N % d == 0:
        D.append(d)
        if d * d != N:
            D.append(N // d)
    d += 1
print(*sorted(D))
```

Делители -1

```python
a,b = map(int,input("input: ").split())

mxsm = 0
sm = 0
for n in range(a,b+1):
    sm = 1 + n
    d = 2
    while d * d <= n:
        if n % d == 0:
            sm += d
            if d * d != n:
                sm += n // d
        d += 1
    if sm >= mxsm:
        mxsm = sm
        mx = n
        

print(mxsm,mx)
```
