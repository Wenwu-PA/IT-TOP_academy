алгоритм поиска

```python
import random

n = int(input("Введите размер массива: "))
A = [random.randint(1, 20) for _ in range(n)]
print("Массив:", *A)

A.sort()
print("После сортировки:", *A)

X = int(input("Введите число X: "))

L = 0
R = len(A) - 1
iterations = 0
found = False

while L <= R:
    iterations += 1

    m1 = L + (R - L) // 3
    m2 = R - (R - L) // 3

    if A[m1] == X or A[m2] == X:
        found = True
        break

    if A[m1] > X:
        R = m1 - 1
    elif A[m2] < X:
        L = m2 + 1
    else:
        L = m1 + 1
        R = m2 - 1
```

расстояние между точками

```python
N, D = map(int, input().split())
A = list(map(int, input().split()))

num = 0
i = 0
j = 0

while i < N and j < N:
    if A[j] - A[i] < D:
        j += 1
    else:
        num += N - j
        i += 1

print(num)
```

количество пар с расстоянием больше либо равно D

```python
N, D = map(int, input().split())
dist = list(map(int, input().split()))

A = [0] * N
for i in range(1, N):
    A[i] = A[i - 1] + dist[i - 1]

num = 0
i = 0
j = 0

while i < N and j < N:
    if A[j] - A[i] < D:
        j += 1
    else:
        num += N - j
        i += 1

print(num)
```
