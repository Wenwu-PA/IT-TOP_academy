линейный поиск

```python
def linear_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1

arr = [1, 2, 3, 4, 5]
linear_search(arr, 3)
```

```python
N = int(input())
A = [int(a) for a in input().split()]
target = int(input())
i = 0
best = A[0]
while i < len(A) and A[i] != target:
    if target - A[i] < best - A[i]:
        best = A[i]
    i += 1
if A[i] == target:
    print(i)
else:
    print(best)
```

бинарный поиск

```python
import random
import math
C = 10
N = [random.randint(1, 100) for _ in range(C)]
print("non sorted: ", N)

sorted_N = sorted(N)

print ("sorted: ", sorted_N)

x = int(input("input a number to search: "))

L = 0; R = len(sorted_N) - 1
i = 0

while L + 1 < R:
    i+=1
    if sorted_N[i] == x:
        print(f"found {x} at index {i}")
        break
    M = (L + R) // 2
    if sorted_N[M] < x:
        L = M
    else:
        R = M
```
