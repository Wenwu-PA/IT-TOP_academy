```python
with open('27_A.txt') as f:
    data = list(map(int, f.read().split()))

N = data[0]
nums = data[1:]

counts = [0] * 100
ans = 0

for x in nums:
    r = x % 100

    ans += counts[r]

    if r != 0 and r != 50:
        ans += counts[100 - r]

    counts[r] += 1

print(ans)
```

```python
with open('27_A.txt') as f:
    data = list(map(int, f.read().split()))

N = data[0]
nums = data[1:]

counts = [0] * 999
ans = 0

for i in range(8, N):
    r = nums[i - 8] % 999
    counts[r] += 1

    r = nums[i] % 999
    ans += counts[r]

print(ans)
```

```python
with open('27-71b.txt') as f:
    data = list(map(int, f.read().split()))

N = data[0]
nums = data[1:]

mx = [-1] * 567
ans = 0

for i in range(11, N):
    x = nums[i - 11]
    r = x % 567
    mx[r] = max(mx[r], x)

    x = nums[i]
    r = x % 567
    need = (567 - r) % 567

    if mx[need] != -1:
        ans = max(ans, x + mx[need])

print(ans)
```
