счастливые номера

```python
N = int(input("input "))

def happyNum(A):
    return A - A//13 - A//17 + A//221

L = 0
R = N * 2
while R >= L:
    R += 2
    m = L + (R - L)//2
    n = happyNum(m)
    if n == N:
        print(m)
        break
    elif n < N:
        L = m + 1
    else:
        R = m - 1
```

методы динамического программирования

```python
K = [1] * (21+1)
K[1] = 1
for i in range(1,21+1):
    K[i] = K[i -1]
    if 1 % 2 == 0:
        K[i] += K[i // 2]
for i in range(11,21):
    K[i] = K[i-1]
    if i%2 == 0 and i//2 >= 10:
        K[i] += K[i//2]
    print(K[21])
```

динамическое программирование

```python
K = [0] * (50+1)
K[4] = 1
for i in range(5,50+1):
    if i == 12:
        continue
    K[i] = K[i -1]
    if i % 3 == 0 and (i <= 6 or i // 3 >= 6):
        K[i] += K[i // 3]
print(K[50])
```

напуганный кузнечик


```python
N = int(input())
Ar1 = []
Ar2 = []
for _ in range(N):
    Ar1.append(int(input()))
for _ in range(N):
    Ar2.append(int(input()))
result = []
for i in range(N):
    result.append(Ar1[i] + Ar2[i])
print(result)
```
