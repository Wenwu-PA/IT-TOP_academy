самый большой квадрат-делитель

```python
x = 594825322637004536
square = 1

p = 2
while p * p <= x:
    cnt = 0
    while x % p == 0:
        x //= p
        cnt += 1

    square *= p ** (cnt // 2 * 2)
    p = 3 if p == 2 else p + 2

print(square)
```

простые числа

```python
def isPrime(n):
    if n % 2 ==0:
        return n == 2
    d = 3
    while d * d <= n and n % d != 0:
        d += 2
    return d * d > n
```

проверка числа на простоту

```python
def isPrime(n):
    if n % 2 ==0:
        return n == 2
    d = 3
    while d * d <= n and n % d != 0:
        d += 2
    return d * d > n
input = int(input("input: "))
if isPrime(input):
    print("Yes")
else:
    print("Not yes")
```

вычисление суммы цифр числа

```python
def sumDigits(n):
    if n < 10:
        return n
    return sumDigits(n // 10) + n % 10
s = sumDigits(1234)
print(s)
```

вывод двоичного кода числа

```python
def printBin(n):
    if n < 1:
        print()
    elif n >= 1:
        printBin(n // 2)
        print(n % 2,end=" ")
printBin(15)
```

примеры рекурсивной функции - суммирование

```python
def sum_array(arr):
    total = 0
    for x in arr:
        total += x
    return total
numbers = [1, 2, 3, 4, 5]
print(sum_array(numbers))
```

задача N ое число Фибоначчи

```python
def fib(n):
    if n <= 2:
        return 1
    else:
        return fib(n - 2) + fib(n - 2)
```

задача на степень

```python
def power(a, n):
    if not isinstance(n, int) or n < 0:
        raise ValueError("Показатель степени должен быть неотрицательным целым числом")

    if n == 0:
        return 1

    half = power(a, n // 2)
    result = half * half

    if n % 2 == 1:
        result *= a

    return result


a = 2
n = 10

print(power(a, n))
```
