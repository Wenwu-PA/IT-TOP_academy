## 1. Мячик на лесенке

```python
n = int(input())

ways = [0] * (n + 1)
ways[0] = 1

for i in range(1, n + 1):
    for step in range(1, 4):
        if i >= step:
            ways[i] += ways[i - step]

print(ways[n])
```

## 2. Стильная одежда

```python
n = int(input())
shirts = list(map(int, input().split()))
m = int(input())
pants = list(map(int, input().split()))

i = 0
j = 0
best = abs(shirts[0] - pants[0])
best_shirt = shirts[0]
best_pants = pants[0]

while i < n and j < m:
    difference = abs(shirts[i] - pants[j])

    if difference < best:
        best = difference
        best_shirt = shirts[i]
        best_pants = pants[j]

    if shirts[i] < pants[j]:
        i += 1
    else:
        j += 1

print(best_shirt, best_pants)
```

## 3. Набор гирек

```python
s, n = map(int, input().split())
weights = sorted(map(int, input().split()))

left = 0
right = n - 1
answer = 0

while left < right:
    total = weights[left] + weights[right]

    if total == s:
        answer += 1
        x = weights[left]
        y = weights[right]

        # Пропускаем одинаковые веса, чтобы не считать их повторно
        while left < right and weights[left] == x:
            left += 1
        while left < right and weights[right] == y:
            right -= 1

    elif total < s:
        left += 1
    else:
        right -= 1

print(answer)
```

## 4. Сбалансированная команда

```python
n = int(input())
skills = sorted(map(int, input().split()))

left = 0
answer = 0

for right in range(n):
    while skills[right] - skills[left] > 5:
        left += 1

    count = right - left + 1
    if count > answer:
        answer = count

print(answer)
```
