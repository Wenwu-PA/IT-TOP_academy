Перевод числа в N-ричную систему

```python
def convert(n, base):
    digits = "0123456789ABCDEF"

    if n >= base:
        convert(n // base, base)

    print(digits[n % base], end="")

n = int(input())
base = int(input())

convert(n, base)
```

Вычисление F(100)

```python
from functools import cache

@cache
def F(n):
    if n <= 3:
        return n

    if n % 3 == 0:
        return n ** 3 + F(n - 1)

    if n % 3 == 1:
        return 4 + F(n // 3)

    return n ** 2 + F(n - 2)

print(F(100))
```

Число Перрена

```python
from functools import cache

@cache
def P(n):
    if n == 0:
        return 3
    if n == 1:
        return 0
    if n == 2:
        return 2

    return P(n - 2) + P(n - 3)

n = int(input())
print(P(n))
```

Исполнитель: из 1 в 402

```python
from functools import cache

@cache
def F(x, even):
    if x > 402 or even > 2:
        return 0

    if x == 402:
        return 1

    result = 0

    for y in (x + 2, x * 2, x * 3):
        count = even

        if y % 2 == 0 and y != 402:
            count += 1

        result += F(y, count)

    return result

print(F(1, 0))
```

Калькулятор BACA

```python
from functools import cache

@cache
def F(x, last):
    if x > 24:
        return 0

    if x == 24:
        return 1

    result = 0

    for cmd, y in (
        ("A", x + 2),
        ("B", x * 2),
        ("C", x * 4)
    ):
        s = last + cmd

        if not s.endswith("BACA"):
            result += F(y, s[-3:])

    return result

print(F(1, ""))
```

Поляков - из 5001 в 45789

```python
start = 5001
end = 45789

dp = [[0] * 4 for i in range(end + 1)]
dp[start][0] = 1

for x in range(start, end):
    for last in range(4):
        if dp[x][last] == 0:
            continue

        commands = [
            (1, x + 1),
            (2, x * 2),
            (3, x + 3)
        ]

        for cmd, y in commands:
            if cmd != last and y <= end:
                dp[y][cmd] += dp[x][last]

print(sum(dp[end]))
```
