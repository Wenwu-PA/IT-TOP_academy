НОД двух чисел

```python
def gcd(a, b):
    while b != 0:
        a, b = b, a % b

    return a


a, b = map(int, input().split())

print(gcd(a, b))
```

Триада Евклида


```python
# def gcd(a, b):
#     while b != 0:
#         a, b = b, a % b

#     return a


def gcd_list(numbers):
    result = numbers[0]

    for i in range(1, len(numbers)):
        result = gcd(result, numbers[i])

    return result


numbers = list(map(int, input().split()))

print(gcd_list(numbers))
```

прямоугольная игра

```python
n = int(input())

answer = n

while n > 1:

    divisor = 2

    while n % divisor != 0:
        divisor += 1

    n = n // divisor

    answer += n

print(answer)
```
