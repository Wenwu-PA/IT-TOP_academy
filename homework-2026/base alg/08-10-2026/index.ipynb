Калькулятор

```python
def cmd2(x):
    # увеличить число десятков на 1; если цифра десятков 9 - число не меняется
    return x if (x // 10) % 10 == 9 else x + 10

memo = {}
def f(x, target=27):
    if x > target:
        return 0
    if x == target:
        return 1
    if x in memo:
        return memo[x]
    memo[x] = f(x + 1, target) + f(cmd2(x), target)
    return memo[x]

def iterative(start=11, target=27):
    dp = [0] * (target + 1)
    dp[start] = 1
    for x in range(start, target):
        dp[x + 1] += dp[x]
        y = cmd2(x)
        if y <= target and y != x:
            dp[y] += dp[x]
    return dp[target]

print(f(11), iterative())
# мне понадобилось почти 2 часа, что бы вникнуть в суть и понять как это можно делать... Надо было сразу после классной работы писать это
```

Ступеньки

```python
#ну вот тут я и присел....
def count_ways(n, steps, banned=0):
    dp = [0] * (n + 1)
    dp[0] = 1# на старте (ступенька 0) один способ - не двигаться
    for i in range(1, n + 1):
        if banned and i % banned == 0:
            continue # запрещённая ступенька: dp[i] остаётся 0
        for s in steps:
            if i >= s:
                dp[i] += dp[i - s]
    return dp[n]

N = int(input())
m, v = map(int, input().split())

# Миша: шаги 1, 2, m; нельзя вставать на кратные 13
misha = count_ways(N, {1, 2, m}, banned=13)

# Вадик: шаги 1, 2, v; нельзя перешагнуть кратную 7, то есть он
# обязательно встаёт на 7, 14, 21... Лестница = блоки по 7 ступенек
q, r = divmod(N, 7)
vadik = count_ways(7, {1, 2, v}) ** q * count_ways(r, {1, 2, v})

if misha > vadik:
    print("Миша", misha - vadik)
elif vadik > misha:
    print("Вадик", vadik - misha)
else:
    print("Ничья", 0)
```
