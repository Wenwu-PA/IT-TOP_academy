## 1. Пароль для Толи

```python
with open("input.txt") as f:
    n = int(f.readline())
    s = f.readline().strip()

k = 0
for c in set(s):
    if s.count(c) % 2 != 0:
        k += 1

if k <= 1:
    ans = "YES"
else:
    ans = "NO"

with open("output.txt", "w") as f:
    f.write(ans)
print(ans)
```

## 2. Совпадение

```python
with open("input.txt") as f:
    n, k = map(int, f.readline().split())

a = n
b = k
while b != 0:
    a, b = b, a % b

ans = n * k // a
with open("output.txt", "w") as f:
    f.write(str(ans))
print(ans)
```

## 3. Импровизированное поле

```python
with open("input.txt") as f:
    m, n = map(int, f.readline().split())

A = [[0] * n for i in range(m)]
for i in range(m):
    A[i][0] = i + 1
for j in range(n):
    A[0][j] = j + 1

for i in range(1, m):
    for j in range(1, n):
        A[i][j] = A[i - 1][j] + A[i][j - 1]

ans = A[m - 1][n - 1]
with open("output.txt", "w") as f:
    f.write(str(ans))
print(ans)
```

## 4. Перестановки

```python
with open("input.txt") as f:
    n = int(f.readline())
    s = f.readline().strip()

k = s.count("1")
a = 1
for i in range(1, k + 1):
    a = a * (n - i + 1) // i

with open("output.txt", "w") as f:
    f.write(str(a))
print(a)
```
