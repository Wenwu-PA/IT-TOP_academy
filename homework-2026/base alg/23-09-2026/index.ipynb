Удаление чисел

```python
n = int(input())
k = int(input())

position = n
count = n
step = 1

while count >= k:
    if position % k == 0:
        print(f"На {step} шаге будет удалено - {n}")
        break
    position -= position // k

    count -= count // k

    step += 1
else:
    print(0)
```

Приближённый бинарный поиск

```python
n, k = map(int, input().split())

a = list(map(int, input().split()))
b = list(map(int, input().split()))


def binary_search(x):
    left = 0
    right = n - 1

    while left <= right:
        mid = (left + right) // 2

        if a[mid] == x:
            return a[mid]
        elif a[mid] < x:
            left = mid + 1
        else:
            right = mid - 1

    if left == 0:
        return a[0]
    elif left == n:
        return a[n - 1]

    left_value = a[left - 1]
    right_value = a[left]

    if x - left_value <= right_value - x:
        return left_value
    else:
        return right_value


for x in b:
    print(binary_search(x))
```

Бинарный поиск медианного вхождения

```python
def lower_bound(a, x):
    left = 0
    right = len(a)

    while left < right:
        mid = (left + right) // 2

        if a[mid] < x:
            left = mid + 1
        else:
            right = mid

    return left


def upper_bound(a, x):
    left = 0
    right = len(a)

    while left < right:
        mid = (left + right) // 2

        if a[mid] <= x:
            left = mid + 1
        else:
            right = mid

    return left


n = int(input())
a = list(map(int, input().split()))

m = int(input())
b = list(map(int, input().split()))

answer = []

for x in b:
    left = lower_bound(a, x)
    right = upper_bound(a, x)

    if left == right:
        answer.append(0)
    else:
        L = left + 1
        R = right

        middle = (L + R) // 2

        answer.append(middle)

print(*answer)
```
