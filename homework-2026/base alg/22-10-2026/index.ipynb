Сумма чисел в массиве

```python
n = int(input())

min_sum = 0
min_index = 0
current_sum = 0

best_sum = -30000 * n - 1
best_i = 0
best_j = 0

for j in range(1, n + 1):
    a = int(input())
    current_sum += a

    segment_sum = current_sum - min_sum
    i = min_index + 1

    if segment_sum > best_sum:
        best_sum = segment_sum
        best_i = i
        best_j = j

    if current_sum <= min_sum:
        min_sum = current_sum
        min_index = j

print(best_i)
print(best_j)
```

Функция F(41)

```python
f = [0] * 42
f[1] = 1

for n in range(2, 42):
    if n % 2 == 1:
        f[n] = n + f[n - 2]
    else:
        f[n] = 2 * n * f[n - 1]

print(f[41])
```

NumberWang

```python
f = [0] * 58
calls = [0] * 58

f[1] = 2
f[2] = 2

calls[1] = 1
calls[2] = 1

for n in range(3, 58):
    f[n] = f[n - 1] - 2 * f[n - 2]
    calls[n] = 1 + calls[n - 1] + calls[n - 2]

print(f[57])
print(calls[57])
```

Корень

```python
n = 4850
count = 0

while int(n ** 0.5) ** 2 != n:
    n += 1
    count += 1

first = int(n ** 0.5) + count

n = 5000
count = 0

while int(n ** 0.5) ** 2 != n:
    n += 1
    count += 1

second = int(n ** 0.5) + count

print(first + second)
```

Функция F(39)

```python
f = [0] * 40
f[1] = 1
f[2] = 1

for n in range(3, 40):
    if n % 2 == 1:
        f[n] = f[n - 1] - f[n - 2]
    else:
        f[n] = sum(f[1:n])

print(f[39])
```

Пять нечётных делителей

```python
numbers = []

for p in range(3, 80, 2):
    prime = True

    for d in range(2, p):
        if p % d == 0:
            prime = False
            break

    if prime:
        number = p ** 4

        while number < 35000000:
            number *= 2

        if number <= 40000000:
            numbers.append(number)

numbers.sort()

for number in numbers:
    print(number)
```
