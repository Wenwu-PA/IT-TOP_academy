Найдите все натуральные числа вида 2
𝑛 × 3
𝑚 × 5
𝑘


```python
numbers = []

for n in range(1, 30, 2):       # n — нечётное
    for m in range(1, 20, 2):   # m — нечётное
        for k in range(1, 15, 2):  # k — нечётное
            x = 2**n * 3**m * 5**k

            if 400_000_000 <= x <= 500_000_000:
                numbers.append(x)

numbers.sort()

print(*numbers, sep="\n")
```

Книги

```python
n, t = map(int, input().split())
a = list(map(int, input().split()))

left = 0
current_sum = 0
answer = 0

for right in range(n):
    current_sum += a[right]

    while current_sum > t:
        current_sum -= a[left]
        left += 1

    answer = max(answer, right - left + 1)

print(answer)
```

Город Че

```python
n, r = map(int, input().split())
d = list(map(int, input().split()))

left = 0
answer = 0

for right in range(n):
    while left < right and d[right] - d[left] <= r:
        left += 1

    answer += left

print(answer)
```

Два дерева на заданном расстоянии

```python
n, D = map(int, input().split())
dist = list(map(int, input().split()))

x = [0]

for d in dist:
    x.append(x[-1] + d)

left = 0

for right in range(n):
    while x[right] - x[left] > D:
        left += 1

    if x[right] - x[left] == D:
        print(left + 1, right + 1)
        break
```
