Провода

```python
n, k = map(int, input().split())

wires = []

for i in range(n):
    wires.append(int(input()))

left = 0
right = max(wires)

while left < right:
    mid = (left + right + 1) // 2

    count = 0

    for wire in wires:
        count += wire // mid

    if count >= k:
        left = mid
    else:
        right = mid - 1

print(left)
```

Ксероксы

```python
n, x, y = map(int, input().split())

first = min(x, y)

left = 0
right = n * max(x, y)

while left < right:
    mid = (left + right) // 2

    copies = mid // x + mid // y

    if copies >= n - 1:
        right = mid
    else:
        left = mid + 1

print(first + left)
```
