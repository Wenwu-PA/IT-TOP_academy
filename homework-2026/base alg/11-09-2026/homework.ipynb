Побеседовав с нейросетью(qwen), я выяснил, что есть еще вариант оптимизации, который заключается в том, что бы избавиться от
остроения полного списка всех делителей числа и можно находить сразу наименьший простой делитель числа, это сокращает кол-во проверок с O(n) до O(sqrt(n))

```python
result = []

for n in range(12345, 12366):
    if n % 2 == 0:
        p = 2
    else:
        p = None
        limit = int(n ** 0.5)
        for d in range(3, limit + 1, 2):
            if n % d == 0:
                p = d
                break
        if p is None:
            continue  # n — простое, делителей нет

    max_divisor = n // p
    if max_divisor <= 4000:
        result.append((n, max_divisor))

print(result)
```

```python
left, right = 12345, 12354
answer = []

for number in range(left, right + 1):
    value = number
    factors = {}
    divisor = 2

    while divisor * divisor <= value:
        while value % divisor == 0:
            factors[divisor] = factors.get(divisor, 0) + 1
            value //= divisor
        divisor = 3 if divisor == 2 else divisor + 2

    if value > 1:
        factors[value] = factors.get(value, 0) + 1

    divisor_count = 1
    for power in factors.values():
        divisor_count *= power + 1

    if divisor_count == 4:
        divisors = [1]
        for prime, power in factors.items():
            divisors = [
                current * prime ** exponent
                for current in divisors
                for exponent in range(power + 1)
            ]

        proper_divisors = [d for d in divisors if d != 1 and d != number]
        answer.append((number, sum(proper_divisors)))

print(answer)
```

Тут он из просто n сделал sqrt(n)

```python
left, right = 12345, 12354
answer = []

for n in range(left, right + 1):
    count = 0
    total = 0
    limit = int(n ** 0.5)
    
    for d in range(2, limit + 1):
        if n % d == 0:
            if d == n // d:
                count += 1
                total += d
            else:
                count += 2
                total += d + n // d
            
            if count > 2:
                break
    
    if count == 2:
        answer.append((n, total))

print(answer)
```

```python
start, end = 100500, 100700

new_answer = []

start_multiple = (start + 26) // 27 * 27

for number in range(start_multiple, end + 1, 27):
    value = number
    p3 = 0
    while value % 3 == 0:
        p3 += 1
        value //= 3
    
    div_count = 1

    if value % 2 == 0:
        power = 0
        while value % 2 == 0:
            power += 1
            value //= 2
        div_count *= (power + 1)
    
    d = 5
    while d * d <= value: #взял из классной работы
        if value % d == 0:
            power = 0
            while value % d == 0:
                power += 1
                value //= d
            div_count *= (power + 1)
            
        if value % (d + 2) == 0:
            power = 0
            while value % (d + 2) == 0:
                power += 1
                value //= (d + 2)
            div_count *= (power + 1)
            
        d += 6
        
    if value > 1:
        div_count *= 2
        
    difference = (p3 - 1) * div_count
    new_answer.append((number, difference))

print(new_answer)
```

тут я решил сначало  найти первое число, кратное на 27, а потом проходиться в цикле только по кратным 27 с шагом 27, потом посчитал сетепени тройки, проверил на делимость и если все гуд, то дальше проверяем делители вида 6 k ± 1 (пропускаем все кратные 2 и 3)

```python
left, right = 123040, 123080

# Начинаем с первого чётного числа
start = left if left % 2 == 0 else left + 1

for n in range(start, right + 1, 2):  # Только чётные числа
    m = n >> 1
    
    # Находим все делители m
    divs = []
    d = 1
    while d * d <= m:
        if m % d == 0:
            divs.append(d)
            if d * d != m:
                divs.append(m // d)
        d += 1
    
    # Если у m ровно 4 делителя, то у n ровно 3 чётных делителя (кроме n)
    if len(divs) == 4:
        # Чётные делители n (кроме n) = 2 * d, где d | m и d != m
        result = sorted(d << 1 for d in divs if d != m)
        print(n, *result)
```
