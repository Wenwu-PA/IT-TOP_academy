Элементы с максимальной разностью

````python
a = list(map(int, input("Элементы массива: ").split()))

if len(a) < 2:
    print("Нужно минимум два элемента")
else:
    min_i = 0
    best_i, best_j = 0, 1
    for j in range(1, len(a)):
        if a[j] - a[min_i] > a[best_j] - a[best_i]:
            best_i, best_j = min_i, j
        if a[j] < a[min_i]:
            min_i = j
    print("Индексы:", best_i + 1, best_j + 1)
    print("Максимальная разность:", a[best_j] - a[best_i])
````

Отношение

````python
n = int(input())
a = list(map(int, input().split()))

min_i = 0
best_i, best_j = -1, -1

for j in range(1, n):
    if a[j] > a[min_i]:
        if best_i == -1 or a[j] * a[best_i] > a[best_j] * a[min_i]:
            best_i, best_j = min_i, j
    if a[j] < a[min_i]:
        min_i = j

if best_i == -1:
    print(0, 0)
else:
    print(best_i + 1, best_j + 1)
````

Дек с защитой от ошибок

````python
from collections import deque

d = deque()
while True:
    parts = input().split()
    command = parts[0]

    if command == "push_front":
        d.appendleft(int(parts[1]))
        print("ok")
    elif command == "push_back":
        d.append(int(parts[1]))
        print("ok")
    elif command == "pop_front":
        print(d.popleft() if d else "error")
    elif command == "pop_back":
        print(d.pop() if d else "error")
    elif command == "front":
        print(d[0] if d else "error")
    elif command == "back":
        print(d[-1] if d else "error")
    elif command == "size":
        print(len(d))
    elif command == "clear":
        d.clear()
        print("ok")
    elif command == "exit":
        print("bye")
        break
#куча элефивов, ужас какой. Кто если не я породит такое
````

Игра в пьяницу (А ведь севе даже играть не надо:( )

````python
from collections import deque

first = deque(map(int, input().split()))
second = deque(map(int, input().split()))

moves = 0
while first and second and moves < 10**6:
    a = first.popleft()
    b = second.popleft()

    if (a == 0 and b == 9) or (a > b and not (a == 9 and b == 0)):
        first.append(a)
        first.append(b)
    else:
        second.append(a)
        second.append(b)
    moves += 1

if not second:
    print("first", moves)
elif not first:
    print("second", moves)
else:
    print("botva")
````
